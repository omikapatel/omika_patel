## 1D NumPy Arrays

```python
list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(list)
arr
```

```python
print(arr*4)
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))

min = (np.argmin(arr))
max = np.argmax(arr)
print(min, max) # gives indexes corresponding to 4 and 50
print(np.min(arr), np.max(arr)) # yes they match

bool_arr = arr[arr>20]
print(bool_arr)
```

## 2D NumPy Arrays

```python
matrix = [
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
]
arr1 = np.array(matrix)

print(np.shape(arr1)) # 3 rows 4 columns
# print(arr1.shape)
print(arr1[:, 2]) # 3rd column
print(arr1[1, :]) # 2nd row

print(arr1.max(axis = 1)) # axis 1 is for all rows, 0 is columns
print(arr1.min(axis = 0))

arr1_flat = arr1.flatten()
print(arr1_flat.shape) # becomes a 12 x 1, all rows merge into one

print(np.argmin(arr1_flat)) # index 7 -> 3 
print(np.argmax(arr1_flat)) # index 5 -> 21

min_coords = np.unravel_index((np.argmin(arr1_flat)), arr1.shape)
max_coords = np.unravel_index((np.argmax(arr1_flat)), arr1.shape)
print(min_coords, max_coords)

print(arr1[arr1>15])

print(np.sort(arr1, axis = 1))
```

## Prime Star Systems

```python
arr2 = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

primes_arr2 = [2, 3, 5, 11, 13, 17]
def alien_life(arr2):
    keep = []
    for row in arr2:
        for value in row:
            if value in primes_arr2:
                keep.append(row)
                break # only one is needed, wont duplicate rows
    return np.array(keep)
print(alien_life(arr2))
            
```

## Let's Play Checkers!

```python
def create_8x8():
    return np.zeros((8,8))
print(create_8x8())

def odd_pattern(grid):
    for row in range(0,8,2):
        for col in range(0,8,2):
            grid[row, col] = 1
    return grid

def even_pattern(grid):
    for row in range(1,8,2):
        for col in range (1,8,2):
            grid[row, col] = 1
    return grid

result = create_8x8()
result = odd_pattern(result)
result = even_pattern(result)
print(result)

reverse = 1-result
print(reverse)
            
```

## The Expanding Universe

```python
def expand(word, n):
    spaced = ""
    for letter in word:
        spaced = spaced + letter + " " * n
    spaced = spaced[:-n]
    return spaced
print(expand("Amazing", 1))

def expansion(universe, n):
    result = []
    for word in universe:
        result.append(expand(word, n))
    return np.array(result)
universe = ["Galaxy", "Cluster"]
print(expansion(universe, 1))
print(expansion(universe, 5)) # lol
```

## Second-Brightest Star

```python
np.random.seed(123)
stars = np.random.randint(500,2000,(5,5))
stars

def second_brightest(stars):
    result = []
    for row in stars:
        sorted_row = np.sort(row)
        result.append(sorted_row[-2])
    return np.array(result)
print(second_brightest(stars))
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Matplotlib

```python
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)
    
x = np.linspace(0, 2*np.pi, 1000)
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
```

```python
fig, ax = plt.subplots(1, 3, figsize=(25, 5))

for A, w, in zip(amplitude, frequency): # zip allows for looping simultaneously
    ax[0].plot(x, make_sine_wave(x, A, 1),
               label=f'A={A}, w=1', linestyle="-")
    ax[0].set_title("Varying Amplitude", fontsize=18)
    ax[0].legend(fontsize=10)
    
    ax[1].plot(x, make_sine_wave(x, 1, w),
               label=f'A=1, w={w}', linestyle="--")
    ax[1].set_title("Varying Frequency", fontsize=18)
    ax[1].legend(fontsize=10)
    ax[2].plot(x, make_sine_wave(x, A, w),
               label=f'A={A}, w={w}', linestyle=":")
    ax[2].set_title("Varying Both", fontsize=18)
    ax[2].legend(fontsize=10)

plt.show()

```

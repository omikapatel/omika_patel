```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("/Users/omikapatel/Desktop/Python_Decal/course_assignments/homework7/star_data.txt", skiprows=1)
print(data.shape) #2642 rows 2 columns
```

```python
magnitude = data[:, 0]
temperature = data[:,1]

print("minimum magnitude:", np.min(magnitude), np.argmin(magnitude)) # brightest
print("maximum magnitude:", np.max(magnitude), np.argmax(magnitude)) # dimmest

print("minimum temperature:", np.min(temperature), np.argmin(temperature))
print("maximum temperature:", np.max(temperature), np.argmax(temperature)) 
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

brightest = np.argmin(magnitude)
dimmest = np.argmax(magnitude)
hottest = np.argmax(temperature)
coldest = np.argmin(temperature)

plt.figure(figsize=(8, 6))
plt.scatter(log_temperature, log_magnitude,
            c=temperature, cmap="plasma",
            s=10, label="All Stars")
plt.colorbar(label="Temperature (K)")

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.scatter(log_temperature[brightest], log_magnitude[brightest],
            color="black", marker="*", s=300, label="Brightest")
plt.scatter(log_temperature[dimmest], log_magnitude[dimmest],
            color="purple", marker="*", s=300, label="Dimmest")
plt.scatter(log_temperature[hottest], log_magnitude[hottest],
            color="green", marker="*", s=300, label="Hottest")
plt.scatter(log_temperature[coldest], log_magnitude[coldest],
            color="orange", marker="*", s=300, label="Coldest")

plt.legend()
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.show()
```

```python
import numpy as np
```

```python
list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(list)
arr
```

## 1D NumPy Arrays

```python
print(arr*4)
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))

min = (np.argmin(arr))
max = np.argmax(arr)
print(min, max) # gives indexes corresponding to 4 and 50
print(np.min(arr), np.max(arr)) # yes they match

```
